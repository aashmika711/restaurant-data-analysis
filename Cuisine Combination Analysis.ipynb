```python
import pandas as pd
df=pd.read_csv("Dataset .csv")
df.columns
```

```python
df = df.dropna(subset=['Cuisines', 'Aggregate rating'])
```

```python
cuisine_counts = df['Cuisines'].value_counts().head(10)
cuisine_counts
```

```python
cuisine_rating = df.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False)
cuisine_rating.head(10)
```

```python
import matplotlib.pyplot as plt
cuisine_counts.plot(kind='bar')
plt.title("Top Cuisine Combinations")
plt.xlabel("Combination")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.show()
```

**Summary**

First, I identified the most common cuisine combinations in the dataset. This helped to understand which types of food are often ordered together.

Next, I calculated the average rating for each cuisine combination. This helped to check whether certain combinations have better ratings than others.

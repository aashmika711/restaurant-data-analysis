```python
import pandas as pd

df = pd.read_csv("Dataset .csv")
df.columns
```

```python
df = df.dropna(subset=['Restaurant Name', 'Aggregate rating', 'Votes'])
```

```python
restaurant_counts = df['Restaurant Name'].value_counts()
restaurant_counts.head()
```

```python
chains = restaurant_counts[restaurant_counts > 1]
chains.head()
```

```python
chain_df = df[df['Restaurant Name'].isin(chains.index)]
```

```python
chain_ratings = chain_df.groupby('Restaurant Name')['Aggregate rating'].mean().sort_values(ascending=False)
```

```python
chain_votes = chain_df.groupby('Restaurant Name')['Votes'].mean().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

chain_ratings.head(10).plot(kind='bar')
plt.title("Top 10 Restaurant Chains by Rating")
plt.xlabel("Restaurant Chain")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```

**Summary**

In this task, I identified restaurant chains by finding restaurant names that appear multiple times in the dataset.

After identifying the chains, I analyzed their average ratings and popularity based on the number of votes.

The results show that some restaurant chains have higher ratings, indicating better quality, while others receive more votes, showing higher popularity among customers.

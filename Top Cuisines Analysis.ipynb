```python
df = df.dropna(subset=['Cuisines'])
```

```python
df['Cuisines'] = df['Cuisines'].str.split(',')
```

```python
cuisines_df = df.explode('Cuisines')
```

```python
cuisines_df['Cuisines'] = cuisines_df['Cuisines'].str.strip()
```

```python
top_cuisines = cuisines_df['Cuisines'].value_counts().head(3)
top_cuisines
```

```python
import pandas as pd
df=pd.read_csv("Dataset .csv")
df.columns
```

```python
df = df.dropna(subset=['Aggregate rating', 'Votes'])
```

```python
rating_counts = df['Aggregate rating'].value_counts().sort_index()
```

```python
import matplotlib.pyplot as plt

rating_counts.plot(kind='bar')
plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
most_common_rating = df['Aggregate rating'].mode()
print("Most common rating:", most_common_rating.values)
```

```python
avg_votes = df['Votes'].mean()
print("Average number of votes:", avg_votes)
```

**Summary**

In this task, I analyzed the ratings and popularity of restaurants in the dataset.

First, I looked at how ratings are distributed across different restaurants. From the analysis, the most common rating is 0, which means many restaurants in the dataset have not received any rating.

Next, I calculated the average number of votes received by restaurants. The average number of votes is 156, which shows that restaurants receive a moderate level of customer engagement.

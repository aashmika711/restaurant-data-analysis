```python
import pandas as pd

df = pd.read_csv("Dataset .csv")
df.columns
```

```python
df = df.dropna(subset=['Has Online delivery', 'Aggregate rating'])
```

```python
delivery_counts = df['Has Online delivery'].value_counts()
delivery_counts
```

```python
total = len(df)
delivery_percentage = (delivery_counts / total) * 100
delivery_percentage
```

```python
avg_rating_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()
avg_rating_delivery
```

```python
import matplotlib.pyplot as plt

delivery_counts.plot(kind='bar')
plt.title("Online Delivery Availability")
plt.xlabel("Online Delivery ")
plt.ylabel("Number of Restaurants")
plt.show()
```

**Summary**

First, I counted the number of restaurants that provide online delivery and those that do not. Then, I calculated the percentage of restaurants in each category.

Next, I compared the average ratings of restaurants with and without online delivery. This helped to understand if online delivery has any impact on customer satisfaction.

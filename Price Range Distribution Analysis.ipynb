```python
import pandas as pd
df = pd.read_csv("Dataset .csv")
```

```python
df.columns
```

```python
df = df.dropna(subset=['Price range'])
```

```python
price_counts = df['Price range'].value_counts().sort_index()
price_counts
```

```python
total = len(df)
price_percentage = (price_counts / total) * 100
price_percentage
```

```python
import matplotlib.pyplot as plt
price_counts.plot(kind='bar')
plt.title("Distribution of Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.show()
```

**Summary:**

First, I counted the number of restaurants in each price range category. Then, I calculated the percentage of restaurants in each category to understand their proportion.

From the analysis, it is clear that some price ranges have more restaurants compared to others. This shows which type of pricing is more common in the dataset.

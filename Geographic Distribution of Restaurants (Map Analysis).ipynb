```python
df.columns
```

```python
df = df.dropna(subset=['Latitude', 'Longitude'])
```

```python
map_center = [df['Latitude'].mean(), df['Longitude'].mean()]
restaurant_map = folium.Map(location=map_center, zoom_start=5)
```

```python
for i, row in df.head(500).iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=row['Restaurant Name']
    ).add_to(restaurant_map)

restaurant_map
```

**Summary**

By plotting the restaurants on a map, I could easily see how they are spread across different areas.

From the map, it is clear that restaurants are grouped in some locations, forming clusters. These clusters show areas where many restaurants are present, while other places have only a few restaurants.
